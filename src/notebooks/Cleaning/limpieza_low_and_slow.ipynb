### Filtrado y limpieza — BENIGN + Low & Slow

Este notebook prepara el conjunto de datos del proyecto: tráfico benigno y ataques lentos (Slowloris y Slowhttptest) de la captura del miércoles 5/7/2017 del dataset CIC-IDS2017 (`GeneratedLabelledFlows`). Se descartan DoS Hulk, DoS GoldenEye y Heartbleed por estar fuera del alcance del protocolo.

Las etapas son: carga, filtrado por clase, exclusión de respuestas del servidor a conexiones de ataque, verificación de hipótesis sobre los datos crudos, limpieza (incluida la corrección del `Timestamp` y la eliminación de columnas sin valor informativo), codificación de la etiqueta binaria, exclusión de ataques con etiqueta incoherente en el tiempo y guardado del resultado.

Toda la lógica reside en `src/preprocessing/clean.py`; este notebook solo la invoca y reporta resultados. Las cifras del resumen final se imprimen desde el código y no se transcriben a mano.

```python
import sys
sys.path.append("../..")  # raíz del repo, desde notebooks/Cleaning/

from pathlib import Path

import pandas as pd

from src.preprocessing.clean import (
    load_raw_csvs,
    filter_low_and_slow,
    excluir_respuestas_de_ataque,
    basic_cleaning,
    encode_labels,
    excluir_ataques_fuera_de_ventana,
)
```

**1. Carga de datos crudos**

`load_raw_csvs` busca recursivamente bajo `data/raw/` el CSV del miércoles que contiene la columna `Timestamp`. Esta columna es necesaria para ordenar los flujos cronológicamente. Deben estar presentes `Flow ID`, `Source IP`, `Destination IP` y `Timestamp`; si no, se cargó una versión del dataset sin esos campos (`MachineLearningCSV`).

```python
df_raw = load_raw_csvs("../../data/raw")
print(f"Shape: {df_raw.shape}")

columnas_esperadas = ["Flow ID", "Source IP", "Destination IP", "Timestamp"]
faltantes = [c for c in columnas_esperadas if c not in df_raw.columns]
assert not faltantes, (
    f"Faltan columnas esperadas {faltantes} -- parece que se cargo el CSV "
    "equivocado (sin Timestamp/Flow ID/IPs). Revisar data/raw/."
)
print("OK: Flow ID, Source IP, Destination IP y Timestamp presentes.")
```

**2. Filtrado: BENIGN + Low & Slow únicamente**

El protocolo se limita a ataques lentos. Se descartan DoS Hulk, DoS GoldenEye y Heartbleed, que son ataques volumétricos o de otra naturaleza, y se conservan el tráfico BENIGN y las clases Slowloris y Slowhttptest.

```python
df_filtrado = filter_low_and_slow(df_raw)
```

**3. Exclusión de respuestas del servidor a conexiones de ataque**

En esta captura los atacantes externos aparecen con la IP de la interfaz NAT del firewall (172.16.0.1) y el servidor web víctima es 192.168.10.50. Todos los flujos NAT → víctima (252,661) están etiquetados como ataque, mientras que los 6,789 flujos BENIGN de ese par van en sentido víctima → NAT. De ellos, 6,787 se emparejan con un flujo de ataque por el puerto del cliente (mediana de la diferencia de inicio: 0 s), y todos ocurren únicamente mientras hay ataques (Slowloris, Slowhttptest, Hulk y GoldenEye). Son el lado servidor de conexiones de ataque que quedó etiquetado como BENIGN, es decir, ruido de etiqueta. Se excluyen y no se reetiquetan, con el mismo criterio que los ataques fuera de ventana (sección 7). La exclusión se hace en este punto porque necesita las IPs, que la limpieza básica elimina.

```python
df_sin_respuestas = excluir_respuestas_de_ataque(df_filtrado)
```

**4. Verificación de hipótesis sobre los datos crudos**

Antes de limpiar, se comprueban sobre `df_filtrado` tres propiedades de los datos que determinan las decisiones de la sección siguiente:

- **Formato del `Timestamp`.** Si las cadenas no contienen AM/PM y las horas parseadas son 1–5 y 8–12 (ninguna de 13 en adelante), la hora está expresada en reloj de 12 horas sin indicador. La captura inicia a las 8:42, de modo que las horas 1–5 corresponden a la tarde. Ordenar con ese formato colocaría la tarde antes que la mañana.
- **Columna duplicada.** `SYN Flag Count` debe coincidir con `Fwd PSH Flags` en el 100% de las filas; en ese caso ambas aportan la misma información y se conserva una sola.
- **Atajos del entorno.** Se mide la proporción de flujos al puerto 80 y la distribución de protocolos por clase. Si una columna separa las clases por la configuración del laboratorio y no por el comportamiento del tráfico, el modelo puede aprovecharla sin detectar realmente un ataque lento.

```python
# Comprueba si el Timestamp trae AM/PM y qué horas aparecen sin corregir
ts_crudo = df_filtrado["Timestamp"].astype(str).str.strip()
print("Cadenas con AM/PM:", int(ts_crudo.str.contains(r"\b[AP]M\b", case=False).sum()))
horas_sin_corregir = pd.to_datetime(ts_crudo, dayfirst=True, errors="coerce").dt.hour
print("Horas tal como vienen en el CSV:")
print(horas_sin_corregir.value_counts().sort_index().to_string())

iguales = (df_filtrado["SYN Flag Count"] == df_filtrado["Fwd PSH Flags"]).mean()
print(f"\nSYN Flag Count == Fwd PSH Flags en {iguales:.4%} de las filas")

clase = df_filtrado["Label"].where(df_filtrado["Label"] == "BENIGN", "ATAQUE")
print("\nProporción de flujos al puerto 80, por clase:")
print((df_filtrado["Destination Port"] == 80).groupby(clase).mean().round(4).to_string())
print("\nProtocolo por clase:")
print(pd.crosstab(df_filtrado["Protocol"], clase, normalize="columns").round(4))
```

**5. Limpieza básica**

La limpieza se aplica con `basic_cleaning` y comprende:

- Conversión del `Timestamp` a fecha y hora (formato D/M/YYYY) y corrección del reloj de 12 horas: las horas menores a 8 se desplazan 12 horas, porque pertenecen a la tarde.
- Reemplazo de infinitos por valores faltantes y eliminación de filas con nulos y de duplicados exactos. Los duplicados se buscan antes de quitar IPs y puertos para eliminar solo copias exactas; dos flujos que difieren únicamente en esas columnas son eventos distintos.
- Eliminación de columnas identificadoras (`Flow ID`, IPs, `Source Port`) para evitar fuga de información.
- Eliminación de atajos del entorno de laboratorio (`Destination Port`, `Protocol`). El 100% de los ataques va al puerto 80 y todos son TCP, mientras que cerca de la mitad del tráfico benigno es UDP; conservar estas columnas permitiría separar las clases sin modelar el comportamiento lento. Para el experimento de ablación "con atajos" se usa `basic_cleaning(df_sin_respuestas, eliminar_atajos=False)` y se guarda el resultado con otro nombre.
- Eliminación de columnas redundantes (incluida `Fwd PSH Flags`, idéntica a `SYN Flag Count`) y de columnas constantes, que no aportan información.
- Eliminación de filas con valores negativos en duración, tiempos entre paquetes o tasas, que no tienen sentido físico.

La celda termina con verificaciones: que el `Timestamp` esté presente y sin valores inválidos, que todas las horas estén dentro del horario de captura (8–17) y que existan flujos de la tarde, lo que confirma que la corrección del reloj se aplicó.

```python
# Limpieza común a los tres modelos y verificación del horario de captura
df_limpio = basic_cleaning(df_sin_respuestas)
print(f"\nShape tras limpieza: {df_limpio.shape}")

assert "Timestamp" in df_limpio.columns, "Timestamp se perdió en basic_cleaning()"
assert df_limpio["Timestamp"].notna().all(), "Hay timestamps no parseables"
horas = df_limpio["Timestamp"].dt.hour
assert horas.between(8, 17).all(), f"Horas fuera de 8-17: {sorted(horas.unique())}"
assert horas.max() >= 13, "No hay flujos de la tarde: la corrección de 12 h no se aplicó"
print("OK: todas las horas están entre 8 y 17 y hay tráfico de la tarde.")
```

**6. Codificación de la etiqueta binaria**

Se crea `label_binaria`: 0 para tráfico BENIGN y 1 para Slowloris o Slowhttptest. La clasificación del proyecto es binaria; la etiqueta original `Label` se conserva para poder reportar el desempeño por tipo de ataque.

```python
df_codificado = encode_labels(df_limpio)
```

**7. Exclusión de ataques fuera de su ventana documentada**

La descripción del dataset documenta las ventanas horarias de cada ataque: Slowloris de 9:47 a 10:10 y Slowhttptest de 10:14 a 10:35. Se aplica una tolerancia de ±2 minutos por la granularidad de un minuto del `Timestamp`.

Slowhttptest queda completamente dentro de su ventana. En Slowloris, seis flujos tienen un `Timestamp` fuera de ella: cinco entre las 14:24 y las 14:25, varias horas después de que terminó el ataque, y uno a las 09:01, antes de que empezara.

Una conexión larga iniciada antes del ataque podría explicar una fila fuera de la ventana. Sin embargo, `Flow Duration` está en microsegundos y estos seis flujos son cortos: cuatro duran alrededor de 5,000,000 µs (5 segundos) y dos duran 229 y 214 µs. Como CICFlowMeter registra el instante de inicio del flujo, ninguno de ellos puede extenderse hasta la ventana del ataque.

Por lo tanto, se consideran ruido de etiquetado y se excluyen. No se reetiquetan como BENIGN porque su clase real es desconocida. La celda imprime las filas excluidas para trazabilidad y se detiene con un error si la exclusión superara el 1% de los ataques, lo que indicaría un problema en el `Timestamp` y no ruido de etiquetado.

```python
# Compara el rango horario de cada ataque antes y después de excluir filas fuera de ventana
print("Rango por ataque ANTES de excluir:")
print(df_codificado[df_codificado["label_binaria"] == 1]
      .groupby("Label")["Timestamp"].agg(["min", "max", "count"]))
print()

df_final = excluir_ataques_fuera_de_ventana(df_codificado)

print("\nRango por ataque DESPUÉS de excluir:")
print(df_final[df_final["label_binaria"] == 1]
      .groupby("Label")["Timestamp"].agg(["min", "max", "count"]))
```

**8. Guardado del dataset procesado**

El resultado se guarda en `data/processed/` y es la entrada común de los notebooks de SVM, XGBoost y GRU. El archivo se regenera localmente ejecutando este notebook sobre la copia de `data/raw/`, cuyo contenido no se modifica.

```python
out_path = Path("../../data/processed/wednesday_low_and_slow_clean.csv")
out_path.parent.mkdir(parents=True, exist_ok=True)
df_final.to_csv(out_path, index=False)
print(f"Guardado: {out_path.resolve()}")
```

**9. Resumen**

La celda siguiente imprime las cifras del procesamiento: filas tras cada etapa, número de columnas, rango del `Timestamp` y distribución de clases. Cualquier cifra que se cite en el informe debe tomarse de esta salida.

Consideraciones para las fases siguientes:

- **Desbalance de clases.** Los ataques representan alrededor del 2.5% de los flujos (relación cercana a 38:1). Debe tratarse con ponderación de clases, submuestreo o SMOTE aplicados solo sobre entrenamiento, y evaluarse con recall, precision, F1 y ROC-AUC en lugar de accuracy.
- **Duplicados.** Los duplicados exactos son pocos porque se buscan antes de quitar IPs y puertos, columnas casi únicas por flujo. Los vectores de features repetidos, que son los que perciben SVM y XGBoost, se cuantifican después de la partición con `reportar_duplicados_de_features()`.
- **Resolución temporal.** El `Timestamp` tiene granularidad de un minuto. Dentro de cada minuto, el orden de los flujos es el del archivo CSV, que no corresponde necesariamente al orden de inicio. El parámetro `desempate` de `train_test_split_by_flow()` permite medir cuánto influye ese orden en el desempeño de la GRU.

```python
features = [c for c in df_final.columns if c not in ("Timestamp", "Label", "label_binaria")]
conteo = df_final["label_binaria"].value_counts()

print(f"Filas crudas:                   {len(df_raw):,}")
print(f"Tras filtro BENIGN + Low&Slow:  {len(df_filtrado):,}")
print(f"Tras excluir respuestas de ataque: {len(df_sin_respuestas):,}")
print(f"Tras limpieza:                  {len(df_limpio):,}")
print(f"Tras excluir fuera de ventana:  {len(df_final):,}")
print(f"\nColumnas del CSV final: {df_final.shape[1]} "
      f"({len(features)} features + Timestamp + Label + label_binaria)")
print(f"Rango de Timestamp: {df_final['Timestamp'].min()} -> {df_final['Timestamp'].max()}")
print(f"\nBENIGN: {conteo[0]:,} | ataque: {conteo[1]:,} | ratio ≈ {conteo[0]/conteo[1]:.1f} : 1")
print(df_final.loc[df_final["label_binaria"] == 1, "Label"].value_counts().to_string())
```
