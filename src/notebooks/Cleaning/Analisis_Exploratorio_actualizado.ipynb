### Análisis exploratorio — BENIGN + Low & Slow

Este notebook describe el dataset procesado por `limpieza_low_and_slow.ipynb` (`data/processed/wednesday_low_and_slow_clean.csv`): tráfico benigno y ataques Slowloris y Slowhttptest del miércoles 5/7/2017 de CIC-IDS2017. **Debe ejecutarse después del notebook de limpieza.**

El análisis se hace sobre el dataset ya limpio: `Timestamp` en reloj de 24 horas, sin identificadores, sin atajos de laboratorio (`Destination Port`, `Protocol`), sin columnas redundantes ni constantes y sin los 6 ataques fuera de ventana. El notebook comprueba primero que el archivo cumpla esas condiciones.

Es un análisis **descriptivo y común a los tres modelos**. Calcula estadísticas sobre el dataset completo para entender los datos, pero **no toma decisiones de modelado**: la selección de features y cualquier ajuste que dependa de la etiqueta se hacen en cada módulo usando solo la partición de entrenamiento, para evitar fuga de información.

Las cifras citadas en el informe deben tomarse de las salidas de este notebook, no transcribirse a mano.

```python
import sys
sys.path.append("../..")  # raíz del repo, desde notebooks/<carpeta>/

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_selection import mutual_info_classif
from sklearn.model_selection import train_test_split

from src.preprocessing.clean import (
    cargar_dataset_limpio,
    VENTANAS_ATAQUE_DOCUMENTADAS,
    COLUMNAS_A_ELIMINAR_CANDIDATAS,
    COLUMNAS_ATAJO_LABORATORIO,
    COLUMNAS_REDUNDANTES,
    COLUMNAS_CONSTANTES,
    HORA_INICIO_CAPTURA,
    HORA_FIN_CAPTURA,
)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
sns.set_style("whitegrid")

RANDOM_STATE = 42
FIG_DIR = Path("../../reports/figures/eda")
FIG_DIR.mkdir(parents=True, exist_ok=True)

# Colores fijos por clase para que todas las figuras sean comparables
COLORES = {"BENIGN": "steelblue", "DoS slowloris": "crimson", "DoS Slowhttptest": "darkorange"}
```

**1. Carga y verificación del dataset**

Se carga con `cargar_dataset_limpio()`, que convierte `Timestamp` a fecha (el CSV lo guarda como texto). Luego se verifica que el archivo corresponda a la versión actual de la limpieza: si alguna comprobación falla, probablemente se está leyendo un CSV generado antes de los últimos cambios de `clean.py` y hay que volver a ejecutar el notebook de limpieza.

```python
file_path = "../../data/processed/wednesday_low_and_slow_clean.csv"
df = cargar_dataset_limpio(file_path)

COLUMNAS_META = ["Timestamp", "Label", "label_binaria"]
features = [c for c in df.columns if c not in COLUMNAS_META]
y = df["label_binaria"]

print(f"Dataset cargado: {df.shape[0]:,} filas x {df.shape[1]} columnas "
      f"({len(features)} features + {len(COLUMNAS_META)} columnas de metadata/etiqueta)")

# Verificaciones de que el CSV corresponde a la limpieza actual
assert pd.api.types.is_datetime64_any_dtype(df["Timestamp"]), "Timestamp no es datetime"
horas = df["Timestamp"].dt.hour
assert horas.between(HORA_INICIO_CAPTURA, HORA_FIN_CAPTURA).all(), "Horas fuera del horario de captura"
assert horas.max() >= 13, "No hay tráfico de la tarde: el CSV no tiene la corrección AM/PM"

eliminadas = (COLUMNAS_A_ELIMINAR_CANDIDATAS + COLUMNAS_ATAJO_LABORATORIO
              + COLUMNAS_REDUNDANTES + COLUMNAS_CONSTANTES)
presentes = [c for c in eliminadas if c in df.columns]
assert not presentes, f"Columnas que la limpieza debió eliminar siguen presentes: {presentes}"

no_numericas = [c for c in features if not pd.api.types.is_numeric_dtype(df[c])]
assert not no_numericas, f"Features no numéricas: {no_numericas}"
assert not df[features].isna().any().any(), "Hay valores nulos en las features"
assert not np.isinf(df[features].to_numpy(dtype=float)).any(), "Hay valores infinitos en las features"

print(f"Rango de Timestamp: {df['Timestamp'].min()} -> {df['Timestamp'].max()}")
print("OK: Timestamp corregido, columnas eliminadas ausentes, sin nulos ni infinitos.")
```

**2. Tipos de datos, columnas constantes y valores negativos**

Se revisa que ninguna feature haya quedado constante y qué columnas conservan valores negativos. La limpieza elimina negativos solo en duración, `Flow IAT Min` y tasas; en otras columnas un negativo puede ser un valor legítimo. Por ejemplo, `Init_Win_bytes_*` = -1 es el valor que CICFlowMeter usa para "no disponible". Esto importa para elegir la transformación logarítmica: `log1p` no admite valores ≤ -1.

```python
print("Tipos de dato de las features:")
print(df[features].dtypes.value_counts().to_string())

constantes = [c for c in features if df[c].nunique() <= 1]
print(f"\nFeatures constantes: {constantes if constantes else 'ninguna'}")

negativos = (df[features] < 0)
cols_con_negativos = negativos.columns[negativos.any()].tolist()
print(f"\nColumnas con valores negativos: {len(cols_con_negativos)}")
if cols_con_negativos:
    resumen_neg = pd.DataFrame({
        "filas_negativas": negativos[cols_con_negativos].sum(),
        "en_ataques": negativos.loc[y == 1, cols_con_negativos].sum(),
        "valor_min": df[cols_con_negativos].min(),
    })
    print(resumen_neg.to_string())
```

**6. Distribuciones por clase**

Histogramas en escala `log1p` con **densidad** (área total = 1 por clase). Con conteos absolutos, la relación 39:1 haría invisible la forma de los ataques. Se incluyen los valores cero (`log1p(0) = 0`), que en la versión anterior se descartaban y ocultaban parte de la distribución.

**8. Asociación de cada feature con la etiqueta**

Se usan tres medidas complementarias:

- **Pearson**: relación lineal. Es sensible a outliers, lo que en estas variables (de 0 a 10⁸) puede distorsionarla.
- **Spearman**: relación monótona, basada en rangos. Es robusta a outliers e igual con o sin transformación logarítmica.
- **Información mutua (MI)**: cualquier tipo de dependencia, incluso no monótona. Por eso es la referencia principal para modelos no lineales como RBF-SVM, XGBoost o GRU. Se estima sobre una muestra estratificada de 50,000 flujos por costo computacional, y sobre las variables en escala logarítmica con signo, porque el estimador es más estable sin colas extremas.

Una correlación baja **no implica** que una feature sea inútil: puede aportar en combinación con otras o de forma no lineal. Esta tabla es descriptiva; **no debe usarse para seleccionar features**, porque se calcula sobre todo el dataset, incluido test.

```python
pearson = df[features].corrwith(y)
spearman = df[features].corrwith(y, method="spearman")

X_mi, _, y_mi, _ = train_test_split(
    df[features], y, train_size=50_000, stratify=y, random_state=RANDOM_STATE
)
X_mi_log = np.sign(X_mi) * np.log1p(np.abs(X_mi))  # log con signo: admite el -1 de Init_Win_bytes_*
# Las columnas con pocos valores (flags 0/1, etc.) se declaran discretas: el estimador
# para variables continuas les asigna una MI pequeña pero positiva aunque no aporten nada
discretas = (X_mi.nunique() <= 10).to_numpy()
X_mi_log.loc[:, discretas] = X_mi.loc[:, discretas]  # las discretas se pasan con sus valores originales
mi = pd.Series(
    mutual_info_classif(X_mi_log, y_mi, discrete_features=discretas, random_state=RANDOM_STATE),
    index=features,
)

asociacion = (
    pd.DataFrame({"pearson": pearson, "spearman": spearman, "info_mutua": mi})
    .sort_values("info_mutua", ascending=False)
)
asociacion.to_csv(FIG_DIR / "asociacion_con_etiqueta.csv")

print("Asociación de cada feature con label_binaria (ordenada por información mutua):")
display(asociacion.round(4))
```

```python
top = asociacion.head(20).iloc[::-1]

fig, axes = plt.subplots(1, 2, figsize=(16, 8), sharey=True)
axes[0].barh(top.index, top["info_mutua"], color="slategray")
axes[0].set_title("Información mutua con la etiqueta (top 20)")
axes[1].barh(top.index, top["spearman"].abs(), color="teal", label="|Spearman|")
axes[1].barh(top.index, top["pearson"].abs(), color="none", edgecolor="black", label="|Pearson|")
axes[1].set_title("|Spearman| y |Pearson| de las mismas features")
axes[1].legend()

plt.tight_layout()
plt.savefig(FIG_DIR / "asociacion_top20.png", dpi=150)
plt.show()
```

**9. Multicolinealidad (diagnóstico)**

Se usa la correlación de **Spearman** entre features, que no depende de la escala ni de los outliers. Se listan los pares con |ρ| > 0.9 y, por separado, los pares con |ρ| > 0.98, que son prácticamente la misma información.

Esta sección **solo diagnostica**. Cada módulo decide qué columnas eliminar con su partición de entrenamiento. Los modelos tienen sensibilidades distintas: XGBoost tolera bien las features correlacionadas, y en SVM o GRU solo conviene eliminar las casi idénticas.

El mapa usa agrupamiento jerárquico, de modo que las features muy correlacionadas aparecen juntas en bloques.

```python
corr_features = df[features].corr(method="spearman")

mask_superior = np.triu(np.ones(corr_features.shape, dtype=bool), k=1)
pares = (
    corr_features.where(mask_superior)
    .stack()
    .rename("rho")
    .reset_index()
    .rename(columns={"level_0": "Feature 1", "level_1": "Feature 2"})
)
pares["abs_rho"] = pares["rho"].abs()
pares_altos = pares[pares["abs_rho"] > 0.9].sort_values("abs_rho", ascending=False)

print(f"Pares con |Spearman| > 0.90: {len(pares_altos)}")
print(f"Pares con |Spearman| > 0.98: {int((pares_altos['abs_rho'] > 0.98).sum())}\n")
with pd.option_context("display.max_rows", None):
    display(pares_altos.drop(columns="abs_rho").round(4).reset_index(drop=True))

pares_altos.drop(columns="abs_rho").to_csv(FIG_DIR / "pares_alta_correlacion_spearman.csv", index=False)
```

```python
g = sns.clustermap(
    corr_features.abs(), cmap="viridis", vmin=0, vmax=1,
    figsize=(18, 18), xticklabels=True, yticklabels=True,
    cbar_kws={"label": "|Spearman|"},
)
g.fig.suptitle("Correlación absoluta entre features (agrupamiento jerárquico)", y=1.01)
g.savefig(FIG_DIR / "clustermap_correlacion.png", dpi=120)
plt.show()
```

**10. Revisión de posibles atajos: `Init_Win_bytes_*`**

`Destination Port` y `Protocol` ya se eliminaron por ser atajos del laboratorio. `Init_Win_bytes_forward` y `Init_Win_bytes_backward` (tamaño inicial de la ventana TCP) pueden comportarse igual: su valor depende sobre todo del sistema operativo y de la herramienta del atacante, no del comportamiento lento del ataque.

Si los ataques se concentran en pocos valores que casi no aparecen en el tráfico benigno, un modelo puede detectarlos solo por esa columna. Esta revisión justifica el experimento de ablación (con y sin estas columnas) previsto en cada módulo.

```python
for col in ["Init_Win_bytes_forward", "Init_Win_bytes_backward"]:
    if col not in df.columns:
        print(f"{col}: no está en el dataset")
        continue

    print(f"=== {col} ===")
    top_valores = (
        df.groupby("Label")[col]
        .value_counts(normalize=True)
        .groupby(level=0)
        .head(3)
        .round(4)
    )
    print("Tres valores más frecuentes por clase (proporción dentro de la clase):")
    print(top_valores.to_string())

    # Qué tan "exclusivos" de los ataques son sus valores
    frecuencia_en_benigno = df.loc[y == 0, col].value_counts(normalize=True)
    frecuencia_de_valor_ataque = df.loc[y == 1, col].map(frecuencia_en_benigno).fillna(0)
    print(f"Ataques cuyo valor aparece en menos del 0.1% del tráfico benigno: "
          f"{(frecuencia_de_valor_ataque < 0.001).mean():.1%}\n")
```

**11. Vectores de features repetidos**

SVM y XGBoost solo ven las features, sin `Timestamp`. Por eso dos flujos con el mismo vector son, para ellos, el mismo ejemplo. Aquí se cuantifica el fenómeno en el dataset completo. Lo que importa para la evaluación, cuántos vectores de val/test ya aparecen en train, se mide después de la partición con `reportar_duplicados_de_features()` de `clean.py`.

Los vectores que aparecen **con ambas etiquetas** indican que, con estas features, esos flujos son indistinguibles: ningún modelo tabular puede clasificarlos todos bien.

```python
hash_vector = pd.util.hash_pandas_object(df[features], index=False)

repetido = hash_vector.duplicated(keep=False)
print("Flujos cuyo vector de features aparece más de una vez:")
print(repetido.groupby(df["Label"]).agg(["sum", "mean"])
      .rename(columns={"sum": "filas", "mean": "proporción"}).round(4).to_string())

etiquetas_por_vector = y.groupby(hash_vector).transform("nunique")
conflictivos = etiquetas_por_vector > 1
print(f"\nFlujos cuyo vector aparece con ambas etiquetas: {int(conflictivos.sum()):,}")
print(df.loc[conflictivos, "Label"].value_counts().to_string())
```

**Implicaciones para el modelado**

Estas conclusiones dependen de la forma de los datos, no de cifras exactas. Contrástalas con las salidas anteriores antes de citarlas en el informe.

- **Escala.** Duraciones, IAT y tasas abarcan varios órdenes de magnitud. SVM y GRU deben usar `scale_features(..., transformacion_log=True)` de `clean.py` (log con signo y luego estandarización), con el mismo valor en ambos módulos. XGBoost no la necesita.
- **Desbalance.** Con una relación cercana a 39:1, la accuracy no es informativa. Conviene usar ponderación de clases o remuestreo solo sobre train, y evaluar con recall, precision, F1, ROC-AUC y PR-AUC.
- **Señal temporal.** Si las features de duración e IAT encabezan la tabla de información mutua y las de volumen quedan abajo, se confirma la premisa del protocolo: el ataque se distingue por sus tiempos y no por la cantidad de tráfico.
- **Multicolinealidad.** Se decide en cada módulo y solo con train. Para SVM y GRU basta eliminar los pares casi idénticos (> 0.98).
- **Atajos.** Si la sección 10 muestra valores de `Init_Win_bytes_*` casi exclusivos de los ataques, hay que reportar los resultados con y sin esas columnas.
- **Vectores conflictivos.** Su número fija un límite a lo que puede lograr un modelo tabular. Si es relevante, es un argumento a favor del enfoque secuencial de la GRU, que ve el contexto de cada flujo.
